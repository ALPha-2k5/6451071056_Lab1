```python
# In thông tin sinh viên
print("Lê Anh Pha - 6451071056")
print("-" * 30)

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Đọc tập dữ liệu Bank Marketing
df = pd.read_csv('bank.csv', sep=';')

# Xem nhanh 5 dòng đầu tiên
display(df.head())
```

Câu 2: Khảo sát sơ bộ dữ liệu

```python
print("Lê Anh Pha - 6451071056")
# Kiểm tra tổng quan thông tin cấu trúc dữ liệu
print("--- Thông tin cấu trúc dữ liệu ---")
df.info()

# Kiểm tra số lượng giá trị thiếu (missing values) trên từng cột
print("\n--- Số lượng giá trị thiếu (Null) ---")
print(df.isnull().sum())
```

Câu 3: Thống kê mô tả các thuộc tính

```python
print("Lê Anh Pha - 6451071056")
# Thống kê mô tả cho các biến định lượng (như Duration, Cons.price.idx)
display(df.describe())

# Thống kê mô tả cho các biến phân loại (như Job, Marital, Education, Subscribe...)
display(df.describe(include=['O']))
```

Câu 4: Tiền xử lý dữ liệu


```python
print("Lê Anh Pha - 6451071056")
# 1. Đối với các cột định lượng (ví dụ: Cons.price.idx), điền giá trị thiếu bằng số trung bình (mean)
if 'Cons.price.idx' in df.columns:
    df['Cons.price.idx'] = df['Cons.price.idx'].fillna(df['Cons.price.idx'].mean())

# 2. Đối với các cột phân loại (ví dụ: Job, Marital...), điền giá trị thiếu bằng giá trị xuất hiện nhiều nhất (mode)
for col in ['Job', 'Marital', 'Education', 'Default', 'Loan', 'Subscribe']:
    if col in df.columns and df[col].isnull().sum() > 0:
        most_common = df[col].mode()[0]
        df[col] = df[col].fillna(most_common)

# Kiểm tra lại sau khi xử lý thiếu
print("Số lượng giá trị thiếu sau khi điền:")
print(df.isnull().sum())
```

Câu 5: Mã hóa dữ liệu phân loại

```python
print("Lê Anh Pha - 6451071056")

# Các cột phân loại (kiểu chữ), trừ cột quyết định y
cat_cols = df.select_dtypes(include='object').columns.drop('y')
print("Các cột phân loại:", list(cat_cols))

# One-hot encoding cho các cột phân loại
df_encoded = pd.get_dummies(df, columns=cat_cols, dtype=int)

# Cột quyết định y: yes -> 1, no -> 0
df_encoded['y'] = df_encoded['y'].map({'yes': 1, 'no': 0})

print(df_encoded.shape)
df_encoded.head()
```
