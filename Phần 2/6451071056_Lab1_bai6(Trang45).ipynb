Câu 1: import và đọc dữ liệu

```python
print("Lê Anh Pha - 6451071056")

import time
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from mlxtend.frequent_patterns import apriori, fpgrowth
from mlxtend.preprocessing import TransactionEncoder

with open('groceries.csv', encoding='utf-8') as f:
    transactions = [[item.strip() for item in line.strip().split(',') if item.strip()]
                    for line in f if line.strip()]

print("Số giao dịch:", len(transactions))
transactions[:5]
```

Câu 2: chuyển sang dạng one-hot

```python
print("Lê Anh Pha - 6451071056")

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(te_ary, columns=te.columns_)
print(basket.shape)
basket.head()
```

Câu 3: Apriori và thời gian xử lý

```python
print("Lê Anh Pha - 6451071056")

#dùng khoảng 10% -> 1% (0.1 -> 0.01) để tránh hết bộ nhớ
min_sup_list = [0.1, 0.08, 0.06, 0.04, 0.02, 0.01]
time_apriori = []
n_apriori = []

for ms in min_sup_list:
    t0 = time.time()
    itemsets_ap = apriori(basket, min_support=ms, use_colnames=True)
    time_apriori.append(time.time() - t0)
    n_apriori.append(len(itemsets_ap))
    print(f"min_sup = {ms*100:.2f}% | {n_apriori[-1]} tập phổ biến | {time_apriori[-1]:.2f} giây")
```

Câu 4: FP-Growth và thời gian xử lý

```python
print("Lê Anh Pha - 6451071056")

time_fp = []
n_fp = []

for ms in min_sup_list:
    t0 = time.time()
    itemsets_fp = fpgrowth(basket, min_support=ms, use_colnames=True)
    time_fp.append(time.time() - t0)
    n_fp.append(len(itemsets_fp))
    print(f"min_sup = {ms*100:.2f}% | {n_fp[-1]} tập phổ biến | {time_fp[-1]:.2f} giây")
```

Câu 5: biểu đồ so sánh thời gian

```python
print("Lê Anh Pha - 6451071056")

x = [ms*100 for ms in min_sup_list]
plt.figure(figsize=(8, 5))
plt.plot(x, time_apriori, marker='o', label='Apriori')
plt.plot(x, time_fp, marker='s', label='FP-Growth')
plt.xlabel('min_sup (%)')
plt.ylabel('Thời gian xử lý (giây)')
plt.title('So sánh thời gian xử lý: Apriori và FP-Growth')
plt.gca().invert_xaxis()   # min_sup giảm dần từ trái sang phải
plt.legend()
plt.grid(True)
plt.show()
```
